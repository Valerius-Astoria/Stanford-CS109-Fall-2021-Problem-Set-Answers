```python
import pandas as pd
```

```python
# Read in the data
data = pd.read_csv('music.csv')
data.head()
```

##### **a.** What is $P(R_{Folk} = 5)$?

$
P(R_{Folk} = 5) = \frac{N(R_{Folk} = 5)}{N(person)} = 0.0564
$

```python
n1 = (data['Folk'] == 5).sum() # N(R_{Folk} = 5)
n2 = len(data) # N(person)
p = n1 / n2
print(f"P(R_Folk) = 5) = {p}")
```

##### **b.** What is $P(R_{Folk} = x)$?

$
P(R_{Folk} = 5) = \frac{N(R_{Folk} = x)}{N(person)}
$

##### **c.** What is $E[R_{Musical}]$

$
E[R_{Musical}] = \frac{\sum R_{Musical}}{N(person)} = 2.7624
$

```python
E = data['Musical'].mean()
print(f"E[R_musical] = {E}")
```

##### **d.** What is $P(R_{Folk} = 5|R_{Musical} = 5)$?

$
P(R_{Folk} = 5|R_{Musical} = 5) = \frac{P(R_{Folk} = 5, R_{Musical} = 5)}{P(R_{Musical} = 5)} = \frac{N(R_{Folk} = 5, R_{Musical} = 5)}{N(R_{Musical} = 5)} = 0.1724
$

```python
n1 = ((data['Folk'] == 5) & (data['Musical'] == 5)).sum()
n2 = (data['Musical'] == 5).sum()
p = n1 / n2
print(f"P(R_Folk=5|R_Musical=5) = {p}")
```

$

##### **f.** What is the covariance of $R_{Opera}$ and $R_{Punk}$?

$
Cov(X, Y) = E[(X - E[X])(Y - E[Y])] = 0.1252
$

```python
print(f"Cov(R_Opera, R_Punk) = {data['Opera'].cov(data['Punk'])}")
```
