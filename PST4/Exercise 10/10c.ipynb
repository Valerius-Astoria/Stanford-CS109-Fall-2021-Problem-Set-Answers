Approximate by sampling

```python
from scipy.stats import beta
import numpy as np
```

```python
def simulate():
    n1 = beta.rvs(3, 4)     # Sample from Beta(3, 4)
    n2 = beta.rvs(13, 4)    # Sample from Beta(13, 4)
    return 1 if n2 > n1 else 0
```

```python
result = np.array([simulate() for _ in range(10_000)])
p = result.sum() / len(result)
print(f"P = {p}")
```
